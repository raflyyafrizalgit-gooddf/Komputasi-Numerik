**Cell 1**

```python
import numpy as np

A = np.array([
    [ 2.0,  3.0, -1.0],
    [ 4.0, -1.0,  2.0],
    [-1.0,  2.0,  3.0]
], dtype=float)

B = np.array([11.0, 7.0, 12.0], dtype=float)

print("Matriks Koefisien A:\n", A)
print("\nVektor Konstanta B:\n", B)
```

**Cell 2**

```python
n = len(B)

M = np.column_stack((A, B))
print("Matriks Augmented Awal:\n", M)

for i in range(n):
    if M[i, i] == 0:
        raise ValueError("Elemen pivot bernilai nol, diperlukan pertukaran baris.")

    for j in range(i + 1, n):
        faktor = M[j, i] / M[i, i]
        M[j, i:] = M[j, i:] - faktor * M[i, i:]

print("\nMatriks Segitiga Atas (Setelah Eliminasi Maju):\n", np.round(M, 4))
```

**Cell 3**

```python
solusi = np.zeros(n)

for i in range(n - 1, -1, -1):
    jumlah_diketahui = np.dot(M[i, i+1:n], solusi[i+1:n])
    solusi[i] = (M[i, n] - jumlah_diketahui) / M[i, i]

print(f"Hasil Penyelesaian SPL: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")
```

**Cell 4**

```python
cek_hasil = np.dot(A, solusi)

print("Hasil pengujian (A * solusi) :", np.round(cek_hasil, 4))
print("Vektor konstanta B asli      :", B)
```
