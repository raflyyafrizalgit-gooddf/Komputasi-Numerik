**Cell 1**

```python
# Mengimpor library NumPy untuk operasi array dan matriks numerik
import numpy as np

# Mendefinisikan matriks koefisien A (3x3) dari sistem persamaan linear baru:
# Baris 1:  2x + 3y -  z = 11
# Baris 2:  4x -  y + 2z = 7
# Baris 3: -1x + 2y + 3z = 12
A = np.array([
    [ 2.0,  3.0, -1.0],
    [ 4.0, -1.0,  2.0],
    [-1.0,  2.0,  3.0]
], dtype=float) #Tipe data float digunakan agar pembagian desimal akurat

# Mendefinisikan vektor konstanta B (ruas kanan persamaan)
B = np.array([11.0, 7.0, 12.0], dtype=float)

# Menampilkan matriks A dan vektor B untuk memastikan input sudah benar
print("Matriks Koefisien A:\n", A)
print("\nVektor Konstanta B:\n", B)
```

Tahap 1: Inisialisasi Library NumPy dan Input Matriks SPL 3 Variabel

**Cell 2**

```python
# Mengambil jumlah baris/variabel dari panjang vektor B (n = 3)
n = len(B)

# Menggabungkan matriks A dan vektor B menjadi matriks augmented M berukuran 3x4
M = np.column_stack((A, B))
print("Matriks Augmented Awal:\n", M)

# Melakukan perulangan untuk setiap kolom pivot i dari 0 hingga n-1
for i in range(n):
    # Memastikan elemen diagonal utama (pivot) tidak bernilai nol
    if M[i, i] == 0:
        raise ValueError("Elemen pivot bernilai nol, diperlukan pertukaran baris.")

    # Melakukan eliminasi pada setiap baris j yang berada di bawah baris pivot i
    for j in range(i + 1, n):
        # Menghitung rasio/faktor pengali untuk mengenolkan elemen di bawah pivot
        faktor = M[j, i] / M[i, i]
        # Mengurangi baris j dengan kelipatan baris pivot i (Operasi Baris Elementer)
        M[j, i:] = M[j, i:] - faktor * M[i, i:]

# Menampilkan matriks segitiga atas hasil eliminasi maju dengan pembulatan 4 desimal
print("\nMatriks Segitiga Atas (Setelah Eliminasi Maju):\n", np.round(M, 4))
```

Tahap 2: Pembentukan Matriks Augmented dan Eliminasi Maju (Forward Elimination)
Pada bagian ini, matriks koefisien `A` digabungkan dengan vektor konstanta `B` membentuk matriks *augmented* `M`. Selanjutnya, dilakukan Operasi Baris Elementer (OBE) untuk mengubah seluruh elemen di bawah diagonal utama (*pivot*) menjadi nol sehingga terbentuk matriks segitiga atas.

**Cell 3**

```python
# Membuat array berisi angka nol berukuran n untuk menyimpan hasil [x, y, z]
solusi = np.zeros(n)

# Melakukan perulangan mundur dari baris terakhir (indeks 2) ke baris pertama (indeks 0)
for i in range(n - 1, -1, -1):
    # Menghitung jumlah perkalian koefisien dan variabel yang sudah ditemukan sebelumnya
    jumlah_diketahui = np.dot(M[i, i+1:n], solusi[i+1:n])
    # Menghitung nilai variabel ke-i dengan membagi sisa konstanta terhadap elemen pivot
    solusi[i] = (M[i, n] - jumlah_diketahui) / M[i, i]

# Menampilkan nilai akhir variabel x, y, dan z ke layar
print(f"Hasil Penyelesaian SPL: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")
```

Tahap 3: Proses Substitusi Balik (Back Substitution)
Setelah matriks berbentuk segitiga atas, perhitungan nilai variabel dilakukan secara mundur dari baris terbawah ke atas—dimulai dari mencari nilai variabel `z`, kemudian menyubstitusikannya ke baris kedua untuk memperoleh `y`, dan terakhir ke baris pertama untuk mendapatkan `x`.

**Cell 4**

```python
# Mengalikan matriks koefisien A dengan vektor solusi [x, y, z] yang diperoleh
cek_hasil = np.dot(A, solusi)

# Menampilkan perbandingan hasil perkalian terhadap vektor konstanta B asli
print("Hasil pengujian (A * solusi) :", np.round(cek_hasil, 4))
print("Vektor konstanta B asli      :", B)
```

Tahap 4: Verifikasi dan Pengujian Akurasi Solusi
Bagian akhir ini menguji ketepatan hasil perhitungan dengan mengalikan kembali matriks koefisien `A` terhadap vektor solusi `(x, y, z)` yang ditemukan, lalu membandingkannya dengan vektor konstanta `B` awal.
