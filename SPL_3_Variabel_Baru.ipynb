**Cell 1**

```python
import numpy as np

# Ganti angka koefisien dan konstanta di bawah ini dengan persamaan unik milik Anda
# Contoh sistem persamaan:
# 3x + 2y -  z = 10
# 2x - 2y + 4z = -2
# -x + 0.5y - z = 0
A = np.array([
    [3.0,  2.0, -1.0],
    [2.0, -2.0,  4.0],
    [-1.0, 0.5, -1.0]
], dtype=float)

B = np.array([10.0, -2.0, 0.0], dtype=float)
print("Matriks Koefisien A:\n", A)
print("Vektor Konstanta B:\n", B)
```

**Cell 2**

```python
n = len(B)
M = np.column_stack((A, B))

for i in range(n):
    if M[i, i] == 0:
        raise ValueError("Pivot bernilai nol, diperlukan pertukaran baris.")
    for j in range(i + 1, n):
        faktor = M[j, i] / M[i, i]
        M[j, i:] = M[j, i:] - faktor * M[i, i:]

print("Matriks setelah Eliminasi Maju:\n", np.round(M, 4))
```

**Cell 3**

```python
solusi = np.zeros(n)

for i in range(n - 1, -1, -1):
    jumlah_diketahui = np.dot(M[i, i+1:n], solusi[i+1:n])
    solusi[i] = (M[i, n] - jumlah_diketahui) / M[i, i]

print(f"Hasil Solusi: x = {solusi[0]:.4f}, y = {solusi[1]:.4f}, z = {solusi[2]:.4f}")
```

**Cell 4**

```python
cek_hasil = np.dot(A, solusi)
print("Hasil perkalian A * solusi :", np.round(cek_hasil, 4))
print("Target vektor B asli      :", B)
```
