```python
import numpy as np

def eliminasi_gauss(A, b):
    # Menggabungkan matriks A dan b menjadi matriks augmented
    n = len(b)
    M = np.concatenate((A, b.reshape(n, 1)), axis=1)

    # 1. Eliminasi Maju (Forward Elimination)
    for i in range(n):
        # Mencegah pembagian dengan nol (Pivoting parsial)
        if M[i, i] == 0.0:
            return "Sistem persamaan tidak memiliki solusi unik atau butuh penukaran baris"

        for j in range(i+1, n):
            rasio = M[j, i] / M[i, i]
            # Mengurangi baris j dengan kelipatan baris i
            M[j, i:] = M[j, i:] - rasio * M[i, i:]

    # 2. Substitusi Mundur (Back Substitution)
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        # Menghitung nilai variabel mulai dari yang paling bawah
        x[i] = (M[i, n] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]

    return x

# ==========================================
# Contoh Penggunaan:
# SPL:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3
# ==========================================

# Matriks koefisien A (wajib menggunakan tipe float)
A = np.array([[2.0, 1.0, -1.0],
              [-3.0, -1.0, 2.0],
              [-2.0, 1.0, 2.0]])

# Vektor hasil b
b = np.array([8.0, -11.0, -3.0])

# Menjalankan fungsi dan mencetak hasil
solusi = eliminasi_gauss(A, b)
print("Matriks A:\n", A)
print("\nVektor b:\n", b)
print("\nSolusi (Nilai x, y, z):")
print(solusi)
```
